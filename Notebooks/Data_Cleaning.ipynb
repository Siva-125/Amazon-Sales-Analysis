## Data Cleaning

```python
df_columns = df.columns
df_columns
```

```python
df1 = df[['Order ID','Date', 'Status', 'Fulfilment', 'Sales Channel ',
          'ship-service-level', 'Category', 'Size', 'Courier Status', 'Qty',
          'Amount', 'ship-city','ship-state', 'B2B', 'fulfilled-by']]
df1.head()
```

```python
df1.isna().sum()
```

```python
df1['Courier Status'] = df['Courier Status'].fillna('Unshipped')
df1['Amount'] = df['Amount'].fillna(0)
df1['ship-city'] = df['ship-city'].fillna('Unknown')
df1['ship-state'] = df['ship-state'].fillna('Unknown')
df1['fulfilled-by'] = df['fulfilled-by'].fillna('Self Ship')

```

```python
df1.isna().sum()
```

```python
df1.info()
```

```python
df1['Date'] = pd.to_datetime(df['Date'])
```

```python
df1['Date'].info()
```

```python
df1['Status'] = df1['Status'].replace('Shipped - Delivered to Buyer','Delivered')
```

```python
df1['Month'] = df1['Date'].dt.month_name()
df1['ship-city'] = df1['ship-city'].str.upper()
df1['ship-state'] = df1['ship-state'].str.upper()
```

```python
df1.columns
```

```python
df1 = df1[['Order ID', 'Date', 'Month', 'Status', 'Fulfilment', 'Sales Channel ',
       'ship-service-level', 'Category', 'Size', 'Courier Status', 'Qty',
       'Amount', 'ship-city', 'ship-state', 'B2B', 'fulfilled-by']]
```

```python
df1.head()
```

- The data is cleaned and transformed to analysis.
- The new attributes (Month) is added.
- The Transformed data is saved on new file.

```python
df1.to_csv('Amazon Sale Report-Transformed.csv', index=False)
```
